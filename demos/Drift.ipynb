```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configuración visual
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# ---------------------------------------------------------
# 1. DATA DRIFT (Covariate Shift) - Cambian los datos de entrada
# Ejemplo: Edad de nuestros clientes
# ---------------------------------------------------------
# Datos de Entrenamiento (ej. Clientes jóvenes)
train_age = np.random.normal(loc=30, scale=5, size=1000)
# Datos de Producción 6 meses después (ej. Expansión a clientes mayores)
prod_age = np.random.normal(loc=45, scale=8, size=1000)

sns.kdeplot(train_age, ax=axes[0], fill=True, label="Entrenamiento (Baseline)")
sns.kdeplot(prod_age, ax=axes[0], fill=True, label="Producción (Actual)")
axes[0].set_title("Data Drift: Cambio en la distribución de la Edad")
axes[0].legend()

# ---------------------------------------------------------
# 2. CONCEPT DRIFT - Cambia la relación entre entrada y salida (Target)
# Ejemplo: Límite de ingresos para aprobar un crédito
# ---------------------------------------------------------
ingresos = np.linspace(1000, 5000, 100)
# Regla en Entrenamiento: Se aprueba si el ingreso > 2000
prob_aprobacion_train = 1 / (1 + np.exp(-(ingresos - 2000) / 100))
# Regla en Producción: Crisis económica, el banco es más estricto, ahora es > 3500
prob_aprobacion_prod = 1 / (1 + np.exp(-(ingresos - 3500) / 100))

axes[1].plot(ingresos, prob_aprobacion_train, label="Política Anterior (Entrenamiento)", color="blue", linewidth=3)
axes[1].plot(ingresos, prob_aprobacion_prod, label="Política Nueva (Producción)", color="red", linestyle="--", linewidth=3)
axes[1].set_title("Concept Drift: Cambio en la probabilidad de Aprobación")
axes[1].set_xlabel("Ingresos del Cliente")
axes[1].set_ylabel("Probabilidad de Aprobación")
axes[1].legend()

plt.tight_layout()
plt.show()

# ---------------------------------------------------------
# Texto Explicativo para Bias y Model Drift
# ---------------------------------------------------------
print("¿Y los otros tipos de Drift?")
print("- MODEL DRIFT: Es la consecuencia de los anteriores. Si el Data Drift o Concept Drift ocurren, el F1-Score o Accuracy de tu modelo caerá (Model Drift).")
print("- BIAS (Sesgo): Ocurre cuando, por ejemplo, el modelo rechaza un 80% más de créditos a una minoría demográfica en Producción respecto a Entrenamiento, aunque la exactitud global se mantenga alta.")
```
